```python
from ultralytics import YOLO
import os

# 2. GPU 사용 가능 여부 확인
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"현재 사용하는 디바이스: {device}")

# 3. 사전 학습된 YOLOv8 가벼운 모델(Nano 버전) 로드
# 처음에는 가볍고 빠른 'yolov8n.pt'로 시작하는 것이 좋습니다.
model = YOLO('yolov8n.pt')

# 4. 모델 학습 실행
# data.yaml 파일의 상대 경로 또는 절대 경로를 정확히 지정해 줍니다.
yaml_path = '주류탐지_데이터셋/data.yaml'

results = model.train(
    data=yaml_path,
    epochs=30,          # 학습 반복 횟수 (상황에 따라 20~50 사이로 조절 가능)
    imgsz=640,          # 이미지 크기 설정
    batch=16,           # 한 번에 처리할 이미지 수 (메모리 부족 시 8로 감소)
    device=device,      # GPU 활용
    name='alcohol_model' # 학습 결과가 저장될 폴더 이름 (runs/detect/alcohol_model)
)

print("=== 모델 학습이 완료되었습니다! ===")
print("저장된 가중치 파일 위치: runs/detect/alcohol_model/weights/best.pt")
```
